```python
from Temperature_loop import TemperatureLoop
from temperature_logger import temperature_logger
from Fake_Cryocon34 import Fakecryocon34

logger = temperature_logger(
    filename="temperature_data.csv",
    channels=("A", "B", "C", "D")
)
cryocon = Fakecryocon34()
```

```python
loop = TemperatureLoop(
    controller=cryocon,
    temperature_arange=[10, 20, 30],
    channels=("A", "B", "C", "D"),
    sample_time=10,
    sampling_frequency=10,
    minimum_offset=1,
    maximum_offset=0.1,
    maximum_std=0.05,
    timeout=60,
    logger=logger
)

```

```python
loop.start()
```

```python
loop.comment("Experiment started")
```

```python
loop.next_temperature()
```

```python
loop.stop()
logger.close()
```
